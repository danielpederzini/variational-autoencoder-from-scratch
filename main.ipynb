```python
import kagglehub
import os
from PIL import Image
import numpy as np
```

```python
dataset_path = kagglehub.dataset_download("splcher/animefacedataset")
dataset_path = os.path.join(dataset_path, "images")

print("Path to dataset files:", dataset_path)
```

```python
image_array = []

for filename in os.listdir(dataset_path):
    if filename.endswith(".jpg") or filename.endswith(".png"):
        image_path = os.path.join(dataset_path, filename)
        image = Image.open(image_path).crop((0, 0, 64, 64))
        image = np.array(image)
        image_array.append(image)

image_array = np.array(image_array)

print("Number of images loaded:", len(image_array))
print("Shape of the image array:", image_array.shape)
print("Data type of the image array:", image_array.dtype)
```
